# 02 — Comparação entre turnos
Usa `eleicoes.analytics.indicators` e `decomposition`. Contagens observadas; decomposições contábeis.

```python
import polars as pl

from eleicoes.analytics.indicators import agregar, variacao_entre_turnos
from eleicoes.config import load_config

cfg = load_config()
GOLD = cfg.paths["gold"]
```

```python
part = pl.read_parquet(GOLD / "gold_participacao_municipio.parquet")
somas = [
    "qt_aptos",
    "qt_comparecimento",
    "qt_abstencoes",
    "qt_votos_brancos",
    "qt_votos_nulos",
    "qt_votos_validos",
]
uf = agregar(part, ["uf", "turno"], somas)
variacao_entre_turnos(uf, ["uf"], ["taxa_comparecimento", "taxa_nulos", "taxa_brancos"])
```

```python
dec = pl.read_parquet(GOLD / "gold_decomposicao_municipio.parquet")
dec.select("uf", "nm_municipio", "contribuicao").sort("contribuicao").head(20)
```
